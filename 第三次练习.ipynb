### 一、基础创建与查看

### 二、列与行操作

1. 选取 score 和 gender 两列。

   

2. 选取索引 0、2 的行。

   

3. 筛选 score>80 的行。

   

4. 新增一列 grade：score≥90 为优秀，否则为良好。

```python
# 1. 选取score和gender两列
df[["score", "gender"]]
```

### 四、分组聚合

### 五、排序与去重

### 六、合并与拼接

1. 横向合并：

   df1 = pd.DataFrame ({"id":[1,2], "name":["X","Y"]})

   df2 = pd.DataFrame ({"id":[1,2], "age":[20,30]})

   按 id 合并。

   

2. 纵向拼接：

   df_a = pd.DataFrame ({"val":[1,2]})

   df_b = pd.DataFrame ({"val":[3,4]})

   拼接后重置索引。


```python
# 1. 按id横向合并
df1 = pd.DataFrame({"id": [1, 2], "name": ["X", "Y"]})
df2 = pd.DataFrame({"id": [1, 2], "age": [20, 30]})
pd.merge(df1, df2, on="id")
```

```python
# 2. 纵向拼接并重置索引
df_a = pd.DataFrame({"val": [1, 2]})
df_b = pd.DataFrame({"val": [3, 4]})
pd.concat([df_a, df_b], ignore_index=True)
```

### 七、时间处理

1. 创建时间列：["2025-01-01","2025-01-02","2025-01-03"]，转为 datetime 格式。

   

2. 提取时间列的月份、星期。


```python
# 1. 转为datetime格式
dates = ["2025-01-01", "2025-01-02", "2025-01-03"]
df_time = pd.DataFrame({"date": pd.to_datetime(dates)})
df_time
```

```python
# 2. 提取月份、星期
df_time["month"] = df_time["date"].dt.month
df_time["weekday"] = df_time["date"].dt.weekday
df_time
```

### 八、高级操作

1. 对 score 列做标准化（(值 - 均值)/ 标准差）。

   

2. 用透视表：行是 gender，值是 score，聚合函数为中位数。

```python
# 1. score列标准化
df["score_std"] = (df["score"] - df["score"].mean()) / df["score"].std()
df[["name", "score", "score_std"]]
```

```python
# 2. 透视表
pd.pivot_table(df, index="gender", values="score", aggfunc="median")
```
